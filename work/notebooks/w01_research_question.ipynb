## 2. The question: decision, action, cost of a wrong call

My question is: **Which pages should an editor review first for refresh or protection?**

- Decision improved: which content pages deserve attention first.
- Unit of analysis: one content page.
- Output: a ranked score and a short reason code for each page.
- Action someone could take: an editor can refresh, expand, or monitor the page.
- Cost of a wrong recommendation: wasted editor time, missed opportunity, and a lower-quality review queue if weak candidates are prioritized.
- Why data or ML can help: the data contains observable signals such as demand, position, engagement, freshness, and trend, which are hard to rank by hand across thousands of pages.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
print('Total rows:', len(df))
print('Rows with impressions_90d >= 500:', int((df['impressions_90d'] >= 500).sum()))
print('Rows with impressions_90d >= 500 and trend_direction == "down":', int(((df['impressions_90d'] >= 500) & (df['trend_direction'] == 'down')).sum()))
print('Rows with impressions_90d >= 500 and sessions_90d >= 30:', int(((df['impressions_90d'] >= 500) & (df['sessions_90d'] >= 30)).sum()))
```
